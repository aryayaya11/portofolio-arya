```python
import pandas
data = pandas.read_csv('data datmin mentah.csv')
```

```python
data
```

# Memisahkan jumlah review dengan rating

```python
for i in range(len(data)):
    mentah = data.loc[i, 'Rating']
    text = mentah.strip()

    rating = text.split(' ')[0]
    jumlah_review = text.split('(')[1].split(' ')[0]

    data.loc[i, 'Rating'] = rating
    data.loc[i, 'Jumlah Review'] = jumlah_review
```

```python
data['Rating'] = data['Rating'].astype(float)
```

# Menyisahkan kecamatan doang untuk alamat serta menghapus kecamatan yang bukan dari surabaya

```python
for i in range(len(data)):
    mentah = data.loc[i, 'Lokasi']
    text = mentah.strip()

    lokasi = text.split(',')[0]

    data.loc[i, 'Lokasi'] = lokasi
```

```python
gk_surabaya = ['Sedati', 'Waru', 'Gedangan', 'Medaeng']
```

```python
didrop = []
for i in range(len(data)):
    if data['Lokasi'][i] in gk_surabaya:
        didrop.append(i)

for i in range(len(didrop)):
    if i == 0:
        data = data.drop(index=didrop[i])
        data = data.reset_index(drop=True)
    else:
        i_real = didrop[i] - i
        data = data.drop(index=i_real)
        data = data.reset_index(drop=True)
```

# Mengubah K dalam jumlah review jadi angka ribuan

```python
for i in range(len(data)):
    text = data['Jumlah Review'][i]
    if ',' in text and 'K' in text:
        hasil = text.replace(',', '.').replace('K', '')
        hasil = int(float(hasil) * 1000)
        data.loc[i, 'Jumlah Review'] = int(hasil)
    elif 'K' in text:
        hasil = text.replace('K', '')
        hasil = int(hasil) * 1000
        data.loc[i, 'Jumlah Review'] = int(hasil)
    else:
        hasil = int(text)
        data.loc[i, 'Jumlah Review'] = int(hasil)

data['Jumlah Review'] = data['Jumlah Review'].astype(int)
```

```python
data.rename(columns={'Lokasi': 'Jarak dari balai kota'}, inplace=True)
```

# Ngubah harga yang awalnya ada Rp nya jadi nominal uang biasa

```python
for i in range(len(data)):
    harga = data['Harga dalam semalam'][i]
    asli = harga.replace('Rp ', '').replace('.', '')
    data.loc[i, 'Harga dalam semalam'] = float(asli)

data['Harga dalam semalam'] = data['Harga dalam semalam'].astype(float)
```

# drop fasilitas yang missing value

```python
data = data.dropna()
data = data.reset_index(drop=True)

for i in range(len(data)):
    angka = data['Jumlah Fasilitas'][i]
    data.loc[i, 'Jumlah Fasilitas'] = int(angka)

data['Jumlah Fasilitas'] = data['Jumlah Fasilitas'].astype(int)
```

# Jadikan CSV

```python
data.to_csv('data_final_datmin.csv', index=False)
```
